# Gikomba Market Stall Performance Analysis

**Analyst:** Fidelmah Mbondo (PLP Data Analytics, 1 Million Devs Cohort)

**Client:** Wanjiku, who runs a fresh-produce stall at Gikomba Market

Wanjiku wrote down her sales for 14 days and keeps a rough record of her stock. She wants to know how her business is doing, whether it is growing, and what she should restock. I use core Python only, with no pandas or NumPy.

## The data

This is the data exactly as Wanjiku gave it to me: 14 days of sales, the day names, and her stock.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Totals, average, highest and lowest days

I use sum(), max() and min() for the totals, and I divide by len() for the average.

To find the day, I use index() to get the position of the best and worst days and then look the name up in days.

Positions 0 to 6 are week 1 and 7 to 13 are week 2, so position // 7 + 1 gives the week number.

```python
# --- Task 1: totals and the best and worst days ---
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest = max(daily_sales)
lowest = min(daily_sales)

# index() gives the position of the first match
best_pos = daily_sales.index(highest)
worst_pos = daily_sales.index(lowest)

best_day = days[best_pos]
worst_day = days[worst_pos]

# positions 0 to 6 are week 1 and 7 to 13 are week 2
best_week = best_pos // 7 + 1
worst_week = worst_pos // 7 + 1

print(f"Total sales:     KES {total_sales:,}")
print(f"Average per day: KES {average_sales:,.2f}")
print(f"Highest day:     KES {highest:,} ({best_day}, week {best_week})")
print(f"Lowest day:      KES {lowest:,} ({worst_day}, week {worst_week})")
```

## Task 2: Week totals and growth

I wrote week_totals() to slice the list into the first 7 days and the last 7 days and return both totals as a tuple.

Then I unpack the tuple into two variables and work out the week-on-week growth as a percentage of week 1.

```python
# --- Task 2: week totals and growth ---
def week_totals(sales):
    """Return the total sales of week 1 and week 2 as a tuple."""
    week_1 = sum(sales[:7])    # first 7 days
    week_2 = sum(sales[7:])    # last 7 days
    return week_1, week_2


week_1_total, week_2_total = week_totals(daily_sales)
growth_pct = (week_2_total - week_1_total) / week_1_total * 100

print(f"Week 1 total: KES {week_1_total:,}")
print(f"Week 2 total: KES {week_2_total:,}")
print(f"Week-on-week growth: {growth_pct:.1f}%")
```

## Task 3: Strong, normal and weak days

classify_day() compares a day with the average.

A day is Strong if it is more than 20% above the average (average * 1.2) and Weak if it is more than 20% below (average * 0.8).

Everything else is Normal. I loop over every day and count the labels with counts.get(label, 0) + 1.

```python
# --- Task 3: strong, normal and weak days ---
def classify_day(amount, average):
    """Return Strong, Normal or Weak by comparing a day with the average."""
    if amount > average * 1.2:       # more than 20% above average
        return "Strong"
    elif amount < average * 0.8:     # more than 20% below average
        return "Weak"
    else:
        return "Normal"


label_counts = {}
for position, amount in enumerate(daily_sales):
    label = classify_day(amount, average_sales)
    label_counts[label] = label_counts.get(label, 0) + 1
    print(f"{days[position]}: KES {amount:,} -> {label}")

print(label_counts)
```

## Task 4: Weekdays against weekends

I loop through the sales and use the day name to put each amount into a weekday list or a weekend list (Sat and Sun). Then I compare the two averages and show how much higher the weekend average is, as a percentage of the weekday average.

```python
# --- Task 4: weekdays against weekends ---
weekday_sales = []
weekend_sales = []

for position, amount in enumerate(daily_sales):
    if days[position] in ("Sat", "Sun"):
        weekend_sales.append(amount)
    else:
        weekday_sales.append(amount)

weekday_avg = sum(weekday_sales) / len(weekday_sales)
weekend_avg = sum(weekend_sales) / len(weekend_sales)

# how much higher the weekend average is, as a percentage of the weekday average
weekend_diff_pct = (weekend_avg - weekday_avg) / weekday_avg * 100

print(f"Weekday average: KES {weekday_avg:,.2f}")
print(f"Weekend average: KES {weekend_avg:,.2f}")
print(f"Weekends are {weekend_diff_pct:.1f}% higher than weekdays")
```

## Task 5: What to reorder and the margins

I loop over the stock dictionary with .items(), find_reorder() returns the products whose quantity is at or below the reorder level, and gross_margin_pct() works out (price - cost) / price as a percentage.

I keep each margin in a dictionary so I can use it in the report.

```python
# --- Task 5: what to reorder and the margins ---
def find_reorder(stock_data):
    """Return the products whose quantity is at or below the reorder level."""
    to_reorder = []
    for name, info in stock_data.items():
        if info["qty"] <= info["reorder"]:
            to_reorder.append(name)
    return to_reorder


def gross_margin_pct(cost, price):
    """Return the gross margin as a percentage of the selling price."""
    return (price - cost) / price * 100


reorder_list = find_reorder(stock)
print("Reorder now:", ", ".join(reorder_list))

margins = {}
for name, info in stock.items():
    margins[name] = gross_margin_pct(info["cost"], info["price"])
    print(f"{name:<9} margin {margins[name]:.1f}%")
```

## Task 6: The Stall Performance Report

This cell pulls the results of Tasks 1 to 5 into one report.

I use f-strings, and a small money() function so every KES amount has thousands separators and two decimals and lines up in the column.

```python
# --- Task 6: the printed report ---
def money(amount):
    """Format an amount as KES with thousands separators."""
    return f"KES {amount:>10,.2f}"


line = "-" * 46

print(line)
print("Stall Performance Report - Gikomba Market")
print(line)

print("Sales (14 days)")
print(f"  Total sales        {money(total_sales)}")
print(f"  Average per day    {money(average_sales)}")
print(f"  Best day           {money(highest)}  ({best_day}, week {best_week})")
print(f"  Worst day          {money(lowest)}  ({worst_day}, week {worst_week})")

print()
print("Week on week")
print(f"  Week 1 total       {money(week_1_total)}")
print(f"  Week 2 total       {money(week_2_total)}")
print(f"  Growth             {growth_pct:>14.1f}%")

print()
print("Types of day")
for label in ["Strong", "Normal", "Weak"]:
    print(f"  {label:<18} {label_counts.get(label, 0):>3} days")

print()
print("Weekdays against weekends")
print(f"  Weekday average    {money(weekday_avg)}")
print(f"  Weekend average    {money(weekend_avg)}")
print(f"  Weekends are {weekend_diff_pct:.1f}% higher")

print()
print("Stock")
print(f"  {'Product':<10}{'Stock':>6}{'Reorder':>9}{'Margin':>9}")
for name, info in stock.items():
    print(f"  {name:<10}{info['qty']:>6}{info['reorder']:>9}{margins[name]:>8.1f}%")
print(f"  Reorder now: {', '.join(reorder_list)}")
print(line)
```
